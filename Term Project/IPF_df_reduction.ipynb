```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv('ipf.csv')

df.info()
```

```python
df.head()
```

```python
df_registration = df[['Name', 'Sex', 'Event', 'Equipment', 'Division', 'WeightClassKg', 'Squat1Kg', 'Bench1Kg', 'Deadlift1Kg']]

df_registration[['Squat1Kg', 'Bench1Kg', 'Deadlift1Kg']] = df_registration[['Squat1Kg', 'Bench1Kg', 'Deadlift1Kg']].abs()

df_registration['ProjectedTotal'] = df_registration[['Squat1Kg', 'Bench1Kg', 'Deadlift1Kg']].sum(axis=1)
df_registration.head()
```

```python
df_registration.size
```

```python
df_registration = df_registration[df_registration['Event'] == 'SBD']
```

```python
df_registration = df_registration.dropna()
df_registration.size
```

```python
df_registration.head()
```

```python
df_registration['Event'].value_counts()
```

```python
df_registration.describe()
```

```python
df_registration = df_registration[df_registration['Squat1Kg'] >= 25]
df_registration = df_registration[df_registration['Bench1Kg'] >= 25]
df_registration = df_registration[df_registration['Deadlift1Kg'] >= 75]

df_registration.describe()
```

```python
plt.plot(df_registration['Squat1Kg'])
plt.show()
```

```python
dfreg_sample = df_registration.sample(n=2000, random_state=42).reset_index().drop('index', axis=1)
dfreg_sample.info()
```

```python
dfreg_sample.to_csv('ipf_reg_2000.csv', index=False)
```
