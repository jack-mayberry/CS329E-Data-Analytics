## Task 1
Summarize the data by whether children participated in the meal preparation or not. Use an

```python
# code here
# getting the data
participants = data['Calorie intake for participants']
non_participants = data['Calorie intake for non-participants']
non_participants = non_participants.dropna()
```

```python
# Calorie intake participants
print(participants.describe())
print()
print(non_participants.describe())
```

```python
# participants
plt.hist(participants)
plt.title('Distribution for Calorie Intake Participants')
plt.show()

fig1, ax1 = plt.subplots()
ax1.set_title('Boxplot of Participants')
ax1.boxplot(participants)
plt.show()
```

```python
# non-participants
plt.hist(non_participants)
plt.title('Distribution for Calorie Intake Non-Participants')
plt.show()

fig2, ax2 = plt.subplots()
ax2.set_title('Boxplot of Non-Participants')
ax2.boxplot(participants)
plt.show()
```

## Task 2
Does the mean calorie consumption for those who participated in the meal preparation differ from 425? Formally test at the α=0.05 level using the 5 steps outlined in the module

```python
# code here
# try out a 2-sided hypothesis t-test
# H_0: µ = 425
# H_1: µ ≠ 425

results = stats.ttest_1samp(participants, 425)

print("T-statistic: ", results.statistic)
print("P-value: ", results.pvalue)
```
